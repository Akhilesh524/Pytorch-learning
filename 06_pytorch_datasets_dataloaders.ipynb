```python
# Dataset

# Understand how training examples are stored and accessed.

# DataLoader

# Learn batching, shuffling, and iteration.


# Training workflow

# Train a model using batches instead of the entire dataset at once.



import torch
import torch.nn as nn
from torch.utils.data import DataLoader

# Model
class SimpleNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)

    def forward(self, x):
        return self.linear(x)


# Dataset
dataset = StudentDataset()

# DataLoader
dataloader = DataLoader(
    dataset,
    batch_size=2,
    shuffle=True
)

# Model, loss and optimizer
torch.manual_seed(42)
model = SimpleNN()

criterion = nn.MSELoss()
optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)

# Training
for epoch in range(100):
    total_loss = 0

    for batch_X, batch_y in dataloader:
        predictions = model(batch_X)

        loss = criterion(predictions, batch_y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    if (epoch + 1) % 20 == 0:
        print(
            f"Epoch {epoch + 1}, "
            f"Loss: {total_loss / len(dataloader):.4f}"
        )

# Test
model.eval()

with torch.no_grad():
    test = torch.tensor([[11.0]])
    prediction = model(test)

print("Prediction for 11:", prediction.item())





```
