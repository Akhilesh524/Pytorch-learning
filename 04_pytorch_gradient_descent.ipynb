```python
# Gradient Descent

# Learn how gradients help us move toward a better solution.

# Loss Function

# Measure how far a prediction is from the correct answer.


# Learning Rate and Parameter Updates

# Understand how much a model changes its parameters during learning.

# Code	Purpose
# loss.backward()	Calculates the gradient.
# torch.no_grad()	Prevents the parameter update from being tracked by Autograd.
# w -= learning_rate * w.grad	Moves the parameter opposite the gradient.
# w.grad.zero_()	Clears the previous gradient before the next iteration.


import torch

# Initial parameter
w = torch.tensor(0.0, requires_grad=True)

# Learning rate
learning_rate = 0.1

# Training loop
for step in range(10):
    loss = (w - 3) ** 2

    loss.backward()

    with torch.no_grad():
        w -= learning_rate * w.grad

    w.grad.zero_()

    print(
        f"Step {step + 1}: "
        f"w = {w.item():.4f}, "
        f"loss = {loss.item():.4f}"
    )
```
