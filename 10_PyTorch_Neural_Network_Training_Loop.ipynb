````python

## Neural Network Training Process & Training Loop

# ### 1. Forward Pass
# Forward pass is the prediction step of a neural network.

# - The model receives input data.
# - Processes it through its layers.
# - Generates predicted output.

# ```python
# prediction = model(X)
# ```

# ### 2. Loss Calculation
# Loss measures the difference between the predicted output and the actual target value.

# - Lower loss generally means better predictions.
# - A loss function measures the model's prediction error.

# Example:

# ```python
# criterion = nn.MSELoss()
# loss = criterion(prediction, y)
# ```

# ### 3. Backpropagation
# Backpropagation calculates gradients of the loss with respect to the model's trainable parameters (weights and biases).

# - It works backward through the network.
# - Calculates gradients using the chain rule.
# - Helps determine how parameters influence the loss.

# ```python
# loss.backward()
# ```

# **Important:** Backpropagation calculates gradients; it does not directly update weights.

# ### 4. Optimizer
# An optimizer updates the model's trainable parameters using the calculated gradients to minimize the loss.

# Example:

# ```python
# optimizer = torch.optim.SGD(
#     model.parameters(),
#     lr=0.01
# )
# ```

# - `SGD`: Stochastic Gradient Descent optimizer.
# - `model.parameters()`: Model's trainable weights and biases.
# - `lr`: Learning rate, which controls the size of parameter updates.

# ### 5. Gradient Clearing – zero_grad()
# PyTorch accumulates gradients by default.

# We use `optimizer.zero_grad()` to clear previous gradients before calculating new ones.

# ```python
# optimizer.zero_grad()
# ```

# ### 6. Backward Pass – backward()
# Calculates gradients for model parameters using automatic differentiation.

# ```python
# loss.backward()
# ```

# ### 7. Parameter Update – optimizer.step()
# Uses calculated gradients to update weights and biases.

# ```python
# optimizer.step()
# ```

# ### 8. Gradient Descent
# Gradient descent updates model parameters in a direction that aims to reduce the loss.

# Formula:

# \[
# w_{new}=w_{old}-\alpha \frac{\partial L}{\partial w}
# \]

# Where:
# - `w` = Weight
# - `α` = Learning rate
# - `L` = Loss
# - `∂L/∂w` = Gradient of loss with respect to weight

# ### 9. Learning Rate
# Learning rate controls how much the optimizer changes model parameters during each update.

# - Too small: Training can be very slow.
# - Appropriate: More stable learning.
# - Too large: Model may overshoot the minimum and fail to converge.

# ### 10. Epoch
# An epoch means one complete pass through the entire training dataset.

# Example:
# - 100 training samples.
# - Epoch 1: All 100 samples are processed once.
# - Epoch 2: All 100 samples are processed again.

# ### 11. Batch
# A batch is a subset of the training dataset processed together in one iteration.

# Example:
# - Dataset size = 100 samples.
# - Batch size = 10.
# - Number of batches = 10.

# ### 12. Iteration
# An iteration is one training-loop execution for one batch.

# Formula:

# \[
# Iterations\ per\ Epoch = \frac{Dataset\ Size}{Batch\ Size}
# \]

# Example:
# - Dataset = 100 samples.
# - Batch size = 10.
# - Iterations per epoch = 10.
# - Epochs = 5.
# - Total iterations = 50.

# ---

# ## Complete Neural Network Training Workflow

# The training process follows these steps:

# 1. Forward Pass – Generate predictions.
# 2. Loss Calculation – Measure prediction error.
# 3. Gradient Clearing – Clear previous gradients.
# 4. Backpropagation – Calculate gradients.
# 5. Optimizer Update – Update weights and biases.
# 6. Repeat the process for multiple epochs.

# ### Basic PyTorch Training Loop

# ```python
# for epoch in range(epochs):

#     # 1. Forward pass
#     prediction = model(X)

#     # 2. Calculate loss
#     loss = criterion(prediction, y)

#     # 3. Clear previous gradients
#     optimizer.zero_grad()

#     # 4. Calculate gradients
#     loss.backward()

#     # 5. Update model parameters
#     optimizer.step()

#     print(f"Epoch: {epoch+1}, Loss: {loss.item():.4f}")
# ```

# ## Important Differences

# | Operation | Purpose |
# |---|---|
# | `optimizer.zero_grad()` | Clears previous gradients |
# | `loss.backward()` | Calculates gradients |
# | `optimizer.step()` | Updates model parameters |
# | `model(X)` | Generates predictions |
# | `criterion()` | Calculates loss |

# ## Key Learnings
# - A neural network learns by adjusting weights and biases.
# - Loss tells us how wrong the predictions are.
# - Backpropagation calculates gradients.
# - The optimizer uses gradients to update parameters.
# - Learning rate controls the update size.
# - Epoch, batch and iteration describe different parts of the training process.

# **Day 10 learning objective:** Understand how a neural network learns through forward propagation, loss calculation, backpropagation and optimizer updates.





import torch
import torch.nn as nn

# Dataset
X = torch.tensor([
    [1.0, 2.0],
    [2.0, 3.0],
    [3.0, 4.0],
    [4.0, 5.0]
])

y = torch.tensor([
    [3.0],
    [5.0],
    [7.0],
    [9.0]
])

# Create model
model = nn.Linear(2, 1)

# Loss function
criterion = nn.MSELoss()

# Optimizer
optimizer = torch.optim.SGD(
    model.parameters(), lr=0.01
)

# Training
epochs = 100

for epoch in range(epochs):

    # Forward pass
    prediction = model(X)

    # Calculate loss
    loss = criterion(prediction, y)

    # Backpropagation
    optimizer.zero_grad()
    loss.backward()

    # Update weights
    optimizer.step()

    if (epoch + 1) % 10 == 0:
        print(
            f"Epoch {epoch+1}, Loss: {loss.item():.4f}"
        )
````
