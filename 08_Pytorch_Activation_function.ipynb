```python
# Exercise 1:Implement ReLU, Sigmoid and Tanh using PyTorch.
import torch
import torch.nn as nn
relu=nn.ReLU()
x = torch.tensor([-5.0, -2.0, 0.0, 2.0, 5.0])
print(relu(x))
sigmoid=nn.Sigmoid()
print(sigmoid(x))
tanh=nn.Tanh()
print(tanh(x))
```

```python
# Exercise 3: Build a neural network with:
# - Input layer: 1 feature.
# - Hidden layer: 8 neurons.
# - Activation: ReLU.
# - Output layer: 1 neuron.
import torch
import torch.nn as nn

class StudentClassifier(nn.Module):

    def __init__(self):
        super().__init__()

        # Hidden layer
        self.layer1 = nn.Linear(1, 8)

        # Activation function
        self.relu = nn.ReLU()

        # Output layer
        self.layer2 = nn.Linear(8, 1)

    def forward(self, x):

        # Input -> Hidden layer
        x = self.layer1(x)

        # Apply activation
        x = self.relu(x)

        # Hidden -> Output layer
        x = self.layer2(x)

        return x


model = StudentClassifier()

# One student, studying 6 hours
input_data = torch.tensor([[6.0]])

logit = model(input_data)

probability = torch.sigmoid(logit)

prediction = (probability >= 0.5).float()

print("Logit:", logit.item())
print("Probability:", probability.item())
print("Prediction:", prediction.item())
```

```python
# Input features: 3

# Hidden neurons: 16

# Activation: ReLU

# Output neurons: 2


import torch
import torch.nn as nn

class NeuralNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(3, 16)
        self.relu = nn.ReLU()
        self.layer2 = nn.Linear(16, 2)

    def forward(self, x):
        x = self.layer1(x)
        x = self.relu(x)
        x = self.layer2(x)
        return x


# Create model outside the class
model = NeuralNetwork()

input_data = torch.tensor([[1.0, 2.0, 3.0]])

output = model(input_data)

print(output)


```
