```python
# day 2
import torch
array = torch.tensor([10,20,30,40])
print("array",array)
print("array shape", array.shape)
print("first element",array[0])
print("last element", array[-1])
```

```python
# 2D tensor

import torch
matrix=torch.tensor([
    [1,2,3],
    [4,5,6]
])
print(matrix)
print(matrix.shape)
print("value of first row and second column", matrix[0,1])
print("value of second row and second column", matrix[1,1])
```

```python
# reshape method for tensor
import torch
x=torch.tensor([1,2,3,4,5,6])
y=x.reshape(3,2)
print(y)
```

```python
# tensor aggregation

import torch
x = torch.tensor([10,20,30,40])
print("sum", x.sum())
print("mean",x.float().mean())
print("max",x.max())
print("min",x.min())
```
